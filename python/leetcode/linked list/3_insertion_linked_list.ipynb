```python
class Node:
    def __init__(self,data):
        self.node=data
        self.next=None

def trav(curr):
    while curr!=None:
        print(curr.node,end=" ")
        curr = curr.next
    print()
    
a = Node(2)
b = Node(3)
c = Node(4)

a.next = b
b.next = c

head = a
trav(head)

new_start = Node(1)
new_start.next = head
head=new_start

trav(head)

new_end = Node(5)

c_node = head
while c_node.next!=None:
    c_node=c_node.next
c_node.next = new_end

trav(head)


mid_node = Node(10)
k = 2
m_c = head
for i in range(k-1):
    m_c=m_c.next
mid_node.next = m_c.next
m_c.next = mid_node

trav(head)
```
