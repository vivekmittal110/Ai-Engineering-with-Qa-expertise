```python
class Node:
    def __init__(self,data):
        self.node = data
        self.next = None

a = Node(4)
b = Node(5)
c = Node(6)

a.next = b
b.next = c
head = a
```

```python
def trav(curr):
    curr=head
    while curr!=None:
        print(curr.node)
        curr = curr.next
```

```python
trav(head)
```

```python
end_n = Node(7)
tn = head
while tn.next != None:
    tn = tn.next    
tn.next = end_n
trav(head)
```

```python
mid_n = Node(8)
k = 2
nn = head
i = 0

while i<k:
    nn = nn.next
    h2 = nn.next
    i+=1
nn = mid_n
mid_n = h2
```
