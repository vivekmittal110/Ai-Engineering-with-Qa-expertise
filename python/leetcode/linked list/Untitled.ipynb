```python
class Node:
    def __init__(self,data):
        self.node = data
        self.next = None

def trav(curr):
    while curr!=None:
        print(curr.node,end="")
        curr = curr.next
    print()
        
a = Node(2)
b = Node(3)
c = Node(4)
d = Node(5)
e = Node(6)
f = Node(7)

a.next = b
b.next = c
c.next = d
d.next = e
e.next = f


head = a
head = head.next

trav(head)

l_node=head
while l_node.next.next!=None:
    l_node=l_node.next
l_node.next = None
trav(head)

m_node=head
k = 2
for i in range(k-1):
    m_node = m_node.next
m_node.next = m_node.next.next
trav(head)
```
